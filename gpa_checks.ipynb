```python
# Author: Corey Williams
# File Name: gpa_checker.py
# Description: This app prompts for student names and GPAs to check if a student 
#              qualifies for the Dean's List (GPA >= 3.5) or the Honor Roll (GPA >= 3.25).
#              It runs continuously until 'ZZZ' is entered as the last name.

def main():
    while True:
        # Prompt for student's last name
        last_name = input("Enter student's last name (or 'ZZZ' to quit): ").strip()
        
        # Check for quit condition
        if last_name.upper() == 'ZZZ':
            print("\nYou've done well, until next time!")
            break
            
        # Prompt for student's first name
        first_name = input("Enter student's first name: ").strip()
        
        # Prompt for student's GPA and validate float input
        while True:
            try:
                gpa = float(input(f"Enter GPA for {first_name} {last_name}: "))
                break
            except ValueError:
                print("Invalid input. Please enter a valid numerical GPA.")

        # Display qualification messages
        if gpa >= 3.5:
            print(f"--> {first_name} {last_name} has made the Dean's List!")
        elif gpa >= 3.25:
            print(f"--> {first_name} {last_name} has made the Honor Roll!")
        else:
            print(f"--> {first_name} {last_name} does not qualify for the Dean's List or Honor Roll.")
            
        print("-" * 40)

if __name__ == "__main__":
    main()
```
